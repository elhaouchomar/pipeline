```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer

column_names = [
    'age', 'menopause', 'tumor-size', 'inv-nodes', 'node-caps',
    'deg-malig', 'breast', 'breast-quad', 'irradiat', 'Class'
]

df = pd.read_csv('data/breast-cancer.csv', names=column_names, header=None)

X = df.drop(columns='Class').dropna()
y = df.loc[X.index, 'Class']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=43, stratify=y
)

ohe_cols = ['node-caps', 'breast', 'breast-quad', 'irradiat']
ord_cols = ["menopause", "age", "tumor-size", "inv-nodes", "deg-malig"]

ord_categories = [
    ['lt40', 'premeno', 'ge40'],
    ['10-19', '20-29', '30-39', '40-49', '50-59', '60-69', '70-79', '80-89', '90-99'],
    ['0-4', '5-9', '10-14', '15-19', '20-24', '25-29', '30-34', '35-39', '40-44', '45-49', '50-54', '55-59'],
    ['0-2', '3-5', '6-8', '9-11', '12-14', '15-17', '18-20', '21-23', '24-26', '27-29', '30-32', '33-35', '36-39'],
    ['1', '2', '3']
]

preprocessor = ColumnTransformer(
    transformers=[
        ('ohe', OneHotEncoder(handle_unknown='ignore', sparse_output=False), ohe_cols),
        ('ord', OrdinalEncoder(categories=ord_categories), ord_cols)
    ],
    remainder='drop'
)

X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed = preprocessor.transform(X_test)

ohe_names = preprocessor.named_transformers_['ohe'].get_feature_names_out(ohe_cols)
feature_names = list(ohe_names) + ord_cols

print("Train shape:", X_train_transformed.shape)
print("Test shape:", X_test_transformed.shape)
print("Feature names:", feature_names[:10])
print(X_test_transformed[:5])
```
